```python
def hungarian(cost):
    n, m = len(cost), len(cost[0])
    INF = float('inf')
    u = [0] * (n + 1)
    v = [0] * (m + 1)
    match = [0] * (m + 1)
    prev = [0] * (m + 1)

    for row in range(1, n + 1):
        match[0] = row
        col = 0
        minv = [INF] * (m + 1)
        used = [False] * (m + 1)

        while match[col] != 0:
            used[col] = True
            r, delta, next_col = match[col], INF, 0
            for j in range(1, m + 1):
                if not used[j]:
                    slack = cost[r - 1][j - 1] - u[r] - v[j]
                    if slack < minv[j]:
                        minv[j], prev[j] = slack, col
                    if minv[j] < delta:
                        delta, next_col = minv[j], j
            for j in range(m + 1):
                if used[j]:
                    u[match[j]] += delta
                    v[j] -= delta
                else:
                    minv[j] -= delta
            col = next_col

        while col:
            match[col] = match[prev[col]]
            col = prev[col]

    assign = [-1] * n
    for j in range(1, m + 1):
        if match[j]:
            assign[match[j] - 1] = j - 1
    return assign

def cost_with_capacity (cost, capacity):  #I added this so that we can have a capacity thing, so like 2 people to one class or hospital or whatever
  seats_available = []
  for c, cap in enumerate(capacities):
    seats_available.extend([c]*cap)

  expanded = [[cost[i][c] for c in seats_available] for i in range(len(cost))]
  return expanded, seats_available

PREFS = {
    "alice": ["math", "physics", "chem"],
    "bob":   ["physics", "chem", "math"],
    "carol": ["math", "chem", "physics"],
    "dan":   ["physics", "chem", "math"],
    "eve":   ["physics", "math", "chem"]
}

agents = list(PREFS)
slots = ["math", "physics", "chem"]
capacities = [2,2,1]
cost = [[PREFS[a].index(s) + 1 for s in slots] for a in agents]

expanded_cost, seats_available = cost_with_capacity(cost, capacities)

for i, j in enumerate(hungarian(expanded_cost)):
    print(agents[i], "->", slots[seats_available[j]])
```
